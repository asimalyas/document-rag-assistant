```python
##create a simple text file

import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_text = {
    "../data/text_files/python_intro.txt": 
    "Python is a high-level, general-purpose programming language known for its simple and readable syntax. It is widely used in web development, data analysis, artificial intelligence, machine learning, automation, and scientific computing.\n\nPython supports different programming concepts such as variables, data types, conditional statements, loops, functions, classes, and modules. Its large collection of libraries makes it useful for many different types of software projects.\n\nFor example, Python can be used to read and process data, build machine learning models, create web applications, automate repetitive tasks, and develop APIs."
}

for filepath,content in sample_text.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)

print("sample file created")
```

```python
##text loader
from langchain_community.document_loaders import TextLoader
loader=TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
doc=loader.load()
print(doc)
```
